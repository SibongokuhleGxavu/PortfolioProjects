# Scraping Data from a Real Website + Pandas

```python
from bs4 import BeautifulSoup
import requests
```

```python
url = 'https://companiesmarketcap.com/south-africa/largest-companies-in-south-africa-by-market-cap/'

page = requests.get(url)

soup = BeautifulSoup(page.text, 'html')
```

```python
print(soup)
```

```python
table = soup.find_all('table', class_ = 'default-table table marketcap-table dataTable')
```

```python
print(table)
```

```python
soup.find('table')
```

```python
table = soup.find_all('table')[0]
```

```python
print(table)
```

```python
sa_titles = table.find_all('th')
```

```python
sa_table_titles = [title.text.strip() for title in sa_titles]
print(sa_table_titles)
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(columns = sa_table_titles)
df
```

```python
column_data = table.find_all('tr')
```

```python
for row in column_data[1:]:
    cells = row.find_all('td')

    row_data = []
    for cell in cells:
        text = cell.get_text(" ", strip=True)   # joins broken lines with spaces
        row_data.append(text)

    if not row_data:
        continue

    if any("Close Ad X" in cell for cell in row_data):
        continue

    if len(row_data) == len(df.columns):
        df.loc[len(df)] = row_data
    else:
        print("Skipped row:", row_data, "Length:", len(row_data))

df
```
